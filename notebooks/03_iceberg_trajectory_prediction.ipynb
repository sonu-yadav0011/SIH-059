```python
# Haversine distance function

def haversine_distance_km(
    lat1,
    lon1,
    lat2,
    lon2
):
    
    # Earth radius in kilometers
    earth_radius_km = 6371.0
    
    
    # Convert degrees to radians
    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)
    
    
    # Coordinate differences
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    
    # Haversine formula
    a = (
        np.sin(dlat / 2) ** 2
        +
        np.cos(lat1)
        *
        np.cos(lat2)
        *
        np.sin(dlon / 2) ** 2
    )
    
    
    c = (
        2
        *
        np.arctan2(
            np.sqrt(a),
            np.sqrt(1 - a)
        )
    )
    
    
    return (
        earth_radius_km
        *
        c
    )
```

```python
# Calculate movement distance

iceberg_tracks["movement_km"] = haversine_distance_km(
    iceberg_tracks["prev_latitude"],
    iceberg_tracks["prev_longitude"],
    iceberg_tracks["Latitude"],
    iceberg_tracks["Longitude"]
)


print(
    iceberg_tracks[
        [
            "Iceberg",
            "date",
            "days_since_previous",
            "movement_km"
        ]
    ].head(15)
)
```

```python
# Convert latitude and longitude
def position_to_grid(
    latitude,
    longitude,
    lat_grid,
    lon_grid
):
    
    # Calculate squared geographic distance
    distance_squared = (
        (lat_grid - latitude) ** 2
        +
        (lon_grid - longitude) ** 2
    )
    
    
    # Find nearest grid cell
    grid_index = np.nanargmin(
        distance_squared
    )
    
    
    # Convert flat index to row and column
    grid_y, grid_x = np.unravel_index(
        grid_index,
        lat_grid.shape
    )
    
    
    return grid_y, grid_x
```

```python
# Convert iceberg position
# to nearest VALID navigation grid cell

def position_to_valid_grid(
    latitude,
    longitude,
    lat_grid,
    lon_grid,
    valid_mask
):
    
    # Find all valid grid cells
    valid_y, valid_x = np.where(
        valid_mask
    )
    
    
    # Get coordinates of valid cells
    valid_latitudes = lat_grid[
        valid_y,
        valid_x
    ]
    
    valid_longitudes = lon_grid[
        valid_y,
        valid_x
    ]
    
    
    # Calculate squared distance
    distance_squared = (
        (valid_latitudes - latitude) ** 2
        +
        (valid_longitudes - longitude) ** 2
    )
    
    
    # Find nearest valid cell
    nearest_index = np.argmin(
        distance_squared
    )
    
    
    # Get original grid coordinates
    grid_y = valid_y[
        nearest_index
    ]
    
    grid_x = valid_x[
        nearest_index
    ]
    
    
    return grid_y, grid_x
```

```python
# Find the ocean current index
# for a given date

def date_to_current_index(
    date,
    current_dates
):
    
    # Convert the input date
    target_date = np.datetime64(
        pd.to_datetime(date).date()
    )
    
    
    # Convert current dates to daily format
    current_dates_daily = np.array(
        current_dates,
        dtype="datetime64[D]"
    )
    
    
    # Find matching date
    matching_indices = np.where(
        current_dates_daily == target_date
    )[0]
    
    
    # Check whether date exists
    if len(matching_indices) == 0:
        
        return None
    
    
    return matching_indices[0]
```

```python
# Get ocean current at
# iceberg position and date

def get_iceberg_ocean_current(
    latitude,
    longitude,
    date,
    lat_grid,
    lon_grid,
    valid_mask,
    current_dates,
    uo,
    vo
):
    
    # Find nearest valid grid cell
    grid_y, grid_x = position_to_valid_grid(
        latitude,
        longitude,
        lat_grid,
        lon_grid,
        valid_mask
    )
    
    
    # Find date index
    time_index = date_to_current_index(
        date,
        current_dates
    )
    
    
    # Check date availability
    if time_index is None:
        
        return None
    
    
    # Retrieve current components
    u_current = uo[
        time_index,
        grid_y,
        grid_x
    ]
    
    v_current = vo[
        time_index,
        grid_y,
        grid_x
    ]
    
    
    return {
        "grid_y": grid_y,
        "grid_x": grid_x,
        "time_index": time_index,
        "u_current": u_current,
        "v_current": v_current
    }
```

```python
# Convert ocean current velocity
# into daily latitude and longitude movement

def current_to_position_change(
    latitude,
    u_current,
    v_current,
    time_seconds=86400
):
    
    # Earth radius in meters
    EARTH_RADIUS_M = 6371000.0
    
    
    # Distance travelled east-west
    distance_east_m = (
        u_current
        *
        time_seconds
    )
    
    
    # Distance travelled north-south
    distance_north_m = (
        v_current
        *
        time_seconds
    )
    
    
    # Convert north-south distance
    # into latitude change
    
    delta_lat_rad = (
        distance_north_m
        /
        EARTH_RADIUS_M
    )
    
    
    delta_lat_deg = np.degrees(
        delta_lat_rad
    )
    
    
    # Convert east-west distance
    # into longitude change
    
    latitude_rad = np.radians(
        latitude
    )
    
    
    delta_lon_rad = (
        distance_east_m
        /
        (
            EARTH_RADIUS_M
            *
            np.cos(latitude_rad)
        )
    )
    
    
    delta_lon_deg = np.degrees(
        delta_lon_rad
    )
    
    
    return (
        delta_lat_deg,
        delta_lon_deg
    )
```

```python
# Simulate iceberg movement using
# daily ocean-current fields

def simulate_iceberg_trajectory(
    start_latitude,
    start_longitude,
    start_date,
    forecast_days,
    lat_grid,
    lon_grid,
    navigation_mask,
    current_dates,
    uo,
    vo
):
    
    current_lat = float(start_latitude)
    current_lon = float(start_longitude)
    
    current_date = pd.Timestamp(start_date)
    
    trajectory = []
    
    
    # Save starting position
    trajectory.append({
        "date": current_date,
        "latitude": current_lat,
        "longitude": current_lon,
        "uo": np.nan,
        "vo": np.nan
    })
    
    
    for day in range(forecast_days):
        
        # Get ocean current for current
        # position and current date
        current_info = get_iceberg_ocean_current(
            latitude=current_lat,
            longitude=current_lon,
            date=current_date,
            lat_grid=lat_grid,
            lon_grid=lon_grid,
            valid_mask=navigation_mask,
            current_dates=current_dates,
            uo=uo,
            vo=vo
        )
        
        
        if current_info is None:
            break
        
        
        u_current = float(
            current_info["u_current"]
        )
        
        v_current = float(
            current_info["v_current"]
        )
        
        
        # Convert current velocity into
        # one day of geographic movement
        delta_lat, delta_lon = (
            current_to_position_change(
                latitude=current_lat,
                u_current=u_current,
                v_current=v_current,
                time_seconds=86400
            )
        )
        
        
        # Update position
        current_lat += delta_lat
        current_lon += delta_lon
        
        
        # Move forward one day
        current_date += pd.Timedelta(days=1)
        
        
        trajectory.append({
            "date": current_date,
            "latitude": current_lat,
            "longitude": current_lon,
            "uo": u_current,
            "vo": v_current
        })
    
    
    return pd.DataFrame(trajectory)
```

```python
# Physics-based trajectory
# with iceberg drift calibration

def simulate_calibrated_trajectory(
    start_latitude,
    start_longitude,
    start_date,
    forecast_days,
    drift_factor,
    lat_grid,
    lon_grid,
    navigation_mask,
    current_dates,
    uo,
    vo
):

    current_lat = float(start_latitude)
    current_lon = float(start_longitude)

    current_date = pd.Timestamp(start_date)

    trajectory = []

    # Starting position
    trajectory.append({
        "date": current_date,
        "latitude": current_lat,
        "longitude": current_lon,
        "uo": np.nan,
        "vo": np.nan
    })

    for day in range(forecast_days):

        current_info = get_iceberg_ocean_current(
            latitude=current_lat,
            longitude=current_lon,
            date=current_date,
            lat_grid=lat_grid,
            lon_grid=lon_grid,
            valid_mask=navigation_mask,
            current_dates=current_dates,
            uo=uo,
            vo=vo
        )

        if current_info is None:
            break

        # Apply drift calibration
        u_current = (
            float(current_info["u_current"])
            * drift_factor
        )

        v_current = (
            float(current_info["v_current"])
            * drift_factor
        )

        # Convert velocity to daily movement
        delta_lat, delta_lon = (
            current_to_position_change(
                latitude=current_lat,
                u_current=u_current,
                v_current=v_current,
                time_seconds=86400
            )
        )

        # Update iceberg position
        current_lat += delta_lat
        current_lon += delta_lon

        # Next day
        current_date += pd.Timedelta(days=1)

        trajectory.append({
            "date": current_date,
            "latitude": current_lat,
            "longitude": current_lon,
            "uo": u_current,
            "vo": v_current
        })

    return pd.DataFrame(trajectory)
```

```python
# Evaluate every drift factor

for drift_factor in drift_factors:

    errors = []

    print(
        "Testing drift factor:",
        drift_factor
    )

    for iceberg_id in moving_iceberg_ids:

        track = (
            moving_iceberg_tracks[
                moving_iceberg_tracks["Iceberg"]
                == iceberg_id
            ]
            .sort_values("date")
            .reset_index(drop=True)
        )

        for i in range(len(track) - 1):

            start_obs = track.iloc[i]
            end_obs = track.iloc[i + 1]

            forecast_days = (
                end_obs["date"]
                -
                start_obs["date"]
            ).days

            if forecast_days <= 0:
                continue

            try:

                trajectory = (
                    simulate_calibrated_trajectory(
                        start_latitude=start_obs["Latitude"],
                        start_longitude=start_obs["Longitude"],
                        start_date=start_obs["date"],
                        forecast_days=forecast_days,
                        drift_factor=drift_factor,
                        lat_grid=lat_model_grid,
                        lon_grid=lon_model_grid,
                        navigation_mask=navigation_mask,
                        current_dates=current_dates,
                        uo=uo,
                        vo=vo
                    )
                )

                predicted_final = trajectory.iloc[-1]

                error_km = haversine_distance_km(
                    predicted_final["latitude"],
                    predicted_final["longitude"],
                    end_obs["Latitude"],
                    end_obs["Longitude"]
                )

                errors.append(error_km)

            except Exception:

                continue


    calibration_results.append({

        "drift_factor":
            drift_factor,

        "mean_error_km":
            np.mean(errors),

        "median_error_km":
            np.median(errors),

        "rmse_km":
            np.sqrt(
                np.mean(
                    np.array(errors) ** 2
                )
            ),

        "predictions":
            len(errors)
    })
```

```python
# Haversine error for persistence baseline


def evaluate_persistence(pairs_df):

    errors = []

    for _, row in pairs_df.iterrows():

        error_km = haversine_distance_km(

            row["start_latitude"],
            row["start_longitude"],

            row["end_latitude"],
            row["end_longitude"]
        )

        errors.append(error_km)


    errors = np.array(errors)


    return {

        "mean_error_km":
            errors.mean(),

        "median_error_km":
            np.median(errors),

        "rmse_km":
            np.sqrt(
                np.mean(errors ** 2)
            ),

        "predictions":
            len(errors)
    }


train_persistence = evaluate_persistence(
    train_pairs
)

test_persistence = evaluate_persistence(
    test_pairs
)


print("TRAIN PERSISTENCE")
print(train_persistence)


print("\nTEST PERSISTENCE")
print(test_persistence)
```

```python
# Evaluate one drift factor
# on a set of observation pairs

def evaluate_drift_factor(
    pairs_df,
    drift_factor,
    lat_grid,
    lon_grid,
    navigation_mask,
    current_dates,
    uo,
    vo
):
    
    errors = []
    
    
    for _, row in pairs_df.iterrows():
        
        try:
            
            trajectory = simulate_calibrated_trajectory(
                
                start_latitude=row["start_latitude"],
                
                start_longitude=row["start_longitude"],
                
                start_date=row["start_date"],
                
                forecast_days=int(
                    row["forecast_days"]
                ),
                
                drift_factor=drift_factor,
                
                lat_grid=lat_grid,
                
                lon_grid=lon_grid,
                
                navigation_mask=navigation_mask,
                
                current_dates=current_dates,
                
                uo=uo,
                
                vo=vo
            )
            
            
            predicted_final = trajectory.iloc[-1]
            
            
            error_km = haversine_distance_km(
                
                predicted_final["latitude"],
                
                predicted_final["longitude"],
                
                row["end_latitude"],
                
                row["end_longitude"]
            )
            
            
            errors.append(error_km)
        
        
        except Exception as error:
            
            print(
                "Skipped:",
                row["Iceberg"],
                row["start_date"],
                error
            )
    
    
    errors = np.array(errors)
    
    
    return {
        
        "drift_factor": drift_factor,
        
        "mean_error_km":
            errors.mean(),
        
        "median_error_km":
            np.median(errors),
        
        "rmse_km":
            np.sqrt(
                np.mean(errors ** 2)
            ),
        
        "predictions":
            len(errors)
    }
```

```python
# 8.62 Iceberg prediction service
# Returns a backend/frontend-ready dictionary


def predict_iceberg_trajectory(
    latitude,
    longitude,
    start_date,
    forecast_days,
    lat_grid,
    lon_grid,
    navigation_mask,
    current_dates,
    uo,
    vo
):
    
    # Convert date
    start_date = pd.Timestamp(start_date)
    
    
    # -----------------------------
    # PERSISTENCE TRAJECTORY MODEL
    # -----------------------------
    
    trajectory = []
    
    
    for day in range(forecast_days + 1):
        
        prediction_date = (
            start_date
            +
            pd.Timedelta(days=day)
        )
        
        
        trajectory.append({
            
            "date":
                prediction_date.strftime(
                    "%Y-%m-%d"
                ),
            
            "latitude":
                float(latitude),
            
            "longitude":
                float(longitude)
        })
    
    
    # --------------------------------
    # GET ENVIRONMENT AT START LOCATION
    # --------------------------------
    
    current_info = get_iceberg_ocean_current(
        
        latitude=latitude,
        
        longitude=longitude,
        
        date=start_date,
        
        lat_grid=lat_grid,
        
        lon_grid=lon_grid,
        
        valid_mask=navigation_mask,
        
        current_dates=current_dates,
        
        uo=uo,
        
        vo=vo
    )
    
    
    # Default environment
    environment = None
    
    
    if current_info is not None:
        
        u_current = float(
            current_info["u_current"]
        )
        
        v_current = float(
            current_info["v_current"]
        )
        
        
        # Calculate speed
        current_speed = float(
            np.sqrt(
                u_current ** 2
                +
                v_current ** 2
            )
        )
        
        
        # Calculate direction
        current_direction = float(
            (
                np.degrees(
                    np.arctan2(
                        u_current,
                        v_current
                    )
                )
                +
                360
            )
            % 360
        )
        
        
        environment = {
            
            "uo_m_per_s":
                u_current,
            
            "vo_m_per_s":
                v_current,
            
            "speed_m_per_s":
                current_speed,
            
            "direction_degrees":
                current_direction,
            
            "grid_y":
                int(
                    current_info["grid_y"]
                ),
            
            "grid_x":
                int(
                    current_info["grid_x"]
                )
        }
    
    
    # -----------------------------
    # FINAL RESPONSE
    # -----------------------------
    
    response = {
        
        "model":
            "persistence_baseline",
        
        "forecast_days":
            int(
                forecast_days
            ),
        
        
        "start_position": {
            
            "latitude":
                float(latitude),
            
            "longitude":
                float(longitude),
            
            "date":
                start_date.strftime(
                    "%Y-%m-%d"
                )
        },
        
        
        "predicted_trajectory":
            trajectory,
        
        
        "environment":
            environment
    }
    
    
    return response
```
